```python
%sql
SELECT (SELECT COUNT(*) FROM rideflow.ldp.trips_bronze) AS bronze,
       (SELECT COUNT(*) FROM rideflow.ldp.trips_silver) AS silver
```

```python
%sql
SELECT COUNT(*) AS rows, COUNT(DISTINCT trip_id) AS trips FROM rideflow.ldp.trips_latest
```

```python
%sql
SELECT COUNT(*) AS total,
       SUM(CASE WHEN __END_AT IS NULL THEN 1 ELSE 0 END) AS current_rows
FROM rideflow.ldp.dim_driver
```

```python
%sql
SELECT COUNT(DISTINCT trip_date_ist) AS days, SUM(total_requests) AS requests
FROM rideflow.ldp.daily_city_kpis
```

```python
%sql
SELECT 'notebook' AS src, trip_date_ist, cancellation_rate_pct FROM rideflow.gold.daily_city_kpis WHERE city_name = 'Mumbai'
UNION ALL
SELECT 'pipeline', trip_date_ist, cancellation_rate_pct FROM rideflow.ldp.daily_city_kpis WHERE city_name = 'Mumbai'
ORDER BY trip_date_ist, src
```
